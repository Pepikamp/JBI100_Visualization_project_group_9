```python
import pandas as pd
from pathlib import Path

# Define file paths
input_path = Path("../data/raw/global_power_plant_database.csv")
output_path = Path("../data/processed/global_power_plant_database.csv")

# Load the original dataset
df = pd.read_csv(input_path)

# Remove the 'url' column
df = df.drop(columns=["url"])
df = df.drop(columns=["name"])
df = df.drop(columns=["owner"])
df = df.drop(columns=["source"])
df = df.drop(columns=["geolocation_source"])
df = df.drop(columns=["generation_data_source"])

# Create the processed folder if it doesn't exist
output_path.parent.mkdir(parents=True, exist_ok=True)

# Save the processed dataset
df.to_csv(output_path, index=False)

print("CSV successfully processed and saved!")
```
